```python
import asyncio
import os
import ssl
from typing import Any, Dict, List
import json
from datetime import datetime
import certifi
from langchain_tavily import TavilyExtract, TavilyMap
from rich.console import Console
from rich.table import Table
from rich.panel import Panel
from rich.progress import Progress
from dotenv import load_dotenv
load_dotenv()
```

```python
ssl_context = ssl.create_default_context(cafile=certifi.where())
os.environ["SSL_CERT_FILE"] = certifi.where()
os.environ["REQUESTS_CA_BUNDLE"] = certifi.where()
```

```python
console = Console()
print("All imports successful ✅")
```

```python
tavily_map = TavilyMap(
    tavily_api_key=os.environ["TAVILY_API_KEY"],
    max_depth=3, # Crawl upto 3 levels deep
    max_breadth=15, # Follow up to 15 links per page
    max_pages=50, # Limit to 50 total pages
) # Essentially creates a sitemap of urls
```

DEMO: Mapping a documentation site

```python
base_url = "https://docs.langchain.com/oss/python/langchain"

console.print(f"Mapping website structure for: {base_url}", style="bold white")

sitemap = tavily_map.invoke(base_url)

urls = sitemap.get("results", [])
console.print(f"Successfully mapped {len(urls)}", style = "bold green")

console.print("First 10 discovered URLs: ", style = "bold yellow")
for i, url in enumerate(urls[:10], 1):
    console.print(f"    {i:2d}. {url}")
```

```python
import random

sample = random.sample(urls, k=2)

console.print(f"Extracting content from 2 random samples", style="bold yellow")
tavily_extract = TavilyExtract()
# ainvoke is asynchronous
extraction_res = await tavily_extract.ainvoke(input={"urls": sample})

# Display results
extracted_docs = extraction_res.get("results", [])
console.print("Extracted sample docs!", style="bold green")

# Show summary of each extracted doc
for i, doc in enumerate(extracted_docs, 1):
    url = doc.get("url", "Unknown")
    content = doc.get("raw_content", "")
    panel_content = (
        f"""URL: {url}\nContent Length: {len(content)}\nPreview: {content}"""
    )
    console.print(Panel(panel_content, title=f"Document {i}", border_style="blue"))
    print()
```

```python
def chunk_urls(urls, chunk_size=3):
    """Split URLs into chunks of specified size"""
    chunks = []
    for i in range(0, len(urls), chunk_size):
        chunks.append(urls[i : i + chunk_size])
    return chunks
```

```python
async def extract_batch(urls, batch_num):
    try:
        console.print(
            f"Processing batch {batch_num} with {len(urls)} URLs", style="bold yellow"
        )
        docs = await tavily_extract.ainvoke(input={"urls": urls})
        res = docs.get("results", [])
        console.print(
            f"Batch {batch_num} completed - extracted {len(res)} documents",
            style="green",
        )
        return res

    except Exception as e:
        console.print(f"Batch {batch_num} FAILED: {e}", style="red")
        return []
```

```python
url_batches = chunk_urls(urls[:9], chunk_size=3)  # Demo of 9
console.print(f"Processing 9 urls in {len(url_batches)} batches", style="bold yellow")

# Process batches concurrently
tasks = [extract_batch(batch, i + 1) for i, batch in enumerate(url_batches)]
batch_results = await asyncio.gather(*tasks)

all = []
for res in batch_results:
    all.extend(res)
with open("tavily_practice_results.json", "w") as f:
    json.dump(all, f)
    
```
